### Problem 1

What is the general solution to the problem $\mathbf{Gm}=\mathbf{d}$, with
$$\begin{bmatrix}
1 & 1 & 0 & 0 \\
0 & 0 & 1 & 1
\end{bmatrix}$$
### Solution

We have 
$$\begin{equation*}
\begin{bmatrix}
1 & 1 & 0 & 0 \\
0 & 0 & 1 & 1
\end{bmatrix}
\begin{bmatrix}
m_1\\
m_2\\
m_3\\
m_4
\end{bmatrix}
=
\begin{bmatrix}
d_1\\
d_2
\end{bmatrix}
\end{equation*}$$

We can see by inspection that both $\mathbf{m}^{par} = [d_1, 0, d_2, 0]^T$ and $\mathbf{m}^{par} = [0, d_1, 0, d_2]^T$ are solutions, let us pick the first one.
Now we need two linearly independant solutions to 

$$\begin{equation*}
\begin{bmatrix}
1 & 1 & 0 & 0 \\
0 & 0 & 1 & 1
\end{bmatrix}
\begin{bmatrix}
m_1\\
m_2\\
m_3\\
m_4
\end{bmatrix}
=
\begin{bmatrix}
0\\
0
\end{bmatrix}
\end{equation*}$$

We can see that $\mathbf{m}^{null(1)} = [1, -1, 0, 0]^T$ and $\mathbf{m}^{null(2)} = [0, 0, 1, -1]^T$ will do the trick
We then have the general solution
$$\begin{equation*}
\mathbf{m}^{gen} =
\begin{bmatrix}
d_1\\
0\\
d_2\\
0
\end{bmatrix}
+ \alpha_1
\begin{bmatrix}
1\\
-1\\
0\\
0
\end{bmatrix}
+ \alpha_2
\begin{bmatrix}
0\\
0\\
1\\
-1
\end{bmatrix}
\end{equation*}$$

### Problem 3

Use SVD on the data kernel $\mathbf{G} = [1, 1, 1, 1]$ 

The null vectors are different from that given in a previous chapter. Why? Show that the two sets are equivalent

### Solution

```python
import numpy as np

G = np.ones((1,4))

U, S, Vh = np.linalg.svd(G)
print("U")
print(U)
print("-"*20)
print("Eigenvalue")
print(S)
print("-"*20)
print("V")
print(Vh)
print("-"*20)
v1 = Vh[1,:]
v2 = Vh[2,:]
v3 = Vh[3,:]
print("The null vectors are")
print("-"*20)
print(v1)
print(v2)
print(v3)
```

The null vectors are different because null vectors are not uniquily determined, and in the earlier chapter they were determined by inspection but here they are generated by an algorithm. Different approches result in different choices of null vectors. They do however span the same space. This can be seen with the transformations. 

$\mathbf{w_i} = - \mathbf{v_i} - \frac{1}{3}\Sigma_j \mathbf{v_j}$ 

$\mathbf{v_i} = - \mathbf{w_i} + \frac{1}{6}\Sigma_j \mathbf{w_j}$

where $\mathbf{w_i}$ are the null vectors from the earlier chapter, $\mathbf{v_i}$ are the SVD null vectors and i and j range from 1 to 3

```python
w1 = -((4/3)*Vh[1,:] + (1/3)*Vh[2,:] + (1/3)*Vh[3,:])
w2 = -((1/3)*Vh[1,:] + (4/3)*Vh[2,:] + (1/3)*Vh[3,:])
w3 = -((1/3)*Vh[1,:] + (1/3)*Vh[2,:] + (4/3)*Vh[3,:])
w1[np.abs(w1) < 1e-15] = 0
w2[np.abs(w2) < 1e-15] = 0
w3[np.abs(w3) < 1e-15] = 0
print(w1)
print(w2)
print(w3)
print("-"*20)
print(-(5/6)*w1 + (1/6)*w2 + (1/6)*w3)
print(-(1/6)*w1 + (5/6)*w2 + (1/6)*w3)
print(-(1/6)*w1 + (1/6)*w2 + (5/6)*w3)
```
